```python
import pandas as pd
```

```python
df = pd.DataFrame({
    'Nombre': ['Ana', 'Luis', 'Carmen'],
    'Edad': [25, 30, 22],
    'Ciudad': ['Madrid', 'Barcelona', 'Valencia']
}, index=['fila1', 'fila2', 'fila3'])

df
```

```python
resultado = df.loc[['fila1','fila3']]
resultado
```
